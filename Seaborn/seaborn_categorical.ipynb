# Seaborn Categorical

```python
import seaborn as sns
import numpy as np
%matplotlib inline
tips=sns.load_dataset('tips')
tips.head()
```

### barplot

```python
sns.barplot(x='sex' ,y='total_bill' ,data=tips)
```

```python
sns.barplot(x='sex' ,y='total_bill' ,data=tips) # defaults to mean
```

```python
sns.barplot(x='sex', y='total_bill', data=tips, estimator=np.std) #standard deviations
```

### countplot

number of occurences

```python
sns.countplot(x='sex',data=tips)
```

### boxplots and violin plots

```python
sns.boxplot(x='day',y='total_bill',data=tips,hue='smoker')
```

Compare boxplot and violin plot

```python
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
# cal also use hue
sns.violinplot(x='day',y='total_bill',data=tips,hue='sex')
```

```python
# most sophisticated
sns.violinplot(x='day',y='total_bill',data=tips,hue='sex',split=True)
```

### Strip Plot

```python
sns.stripplot(x='day',y='total_bill',data=tips)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=True)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,jitter=True,hue='sex')
```

#### Swarmplot
doesn't stack points on top of eachother, very similar to violin plot; 
don't use these for final evals, use them for exploratory

```python
sns.swarmplot(x='day',y='total_bill',data=tips)
```

#### Factor plot
an alternative to calling a specific plot, uses a `kind` variable to specify the type of plot. 

```python
sns.factorplot(x='day',y='total_bill',data=tips,kind='bar')
```
