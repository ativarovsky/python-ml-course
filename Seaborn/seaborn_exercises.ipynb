# Seaborn Exercises

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic=sns.load_dataset('titanic')
```

```python
titanic.head()
```

#### Exercise 1

```python
sns.jointplot(x='fare',y='age',data=titanic, kind='scatter')
```

#### Exercise 2

```python
sns.histplot(titanic['fare'], bins=30, color='red')
```

#### Exercise 3

boxplot of age by class

```python
sns.boxplot(x='class',y='age',data=titanic)
```

```python
sns.swarmplot(x='class', y='age', data=titanic, hue='class')
```

```python
sns.countplot(x='sex',data=titanic,hue='sex')
```

```python
sns.histplot(x='age', data=titanic)
```
