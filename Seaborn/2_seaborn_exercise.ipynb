## Seaborn Exercises 2

```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

Create jointplot

```python
sns.jointplot(x='fare', y='age', data=titanic, kind='scatter')
```

Create histogram of fare 

```python
sns.displot(titanic['fare'], bins=30, color='red')
plt.margins(0)
```

Boxplot of class by age

```python
sns.boxplot(x='class', y='age', data=titanic, palette='rainbow')
plt.margins(0)
```

Create swarm plot

```python
sns.swarmplot(x='class',y='age',data=titanic, palette='Set2', size=4)
```

Barplot of sex

```python
sns.countplot(x='sex',data=titanic)
```

Heatmap of correlations of variables

```python
tc=titanic.corr()
```

```python
sns.heatmap(data=tc,cmap='coolwarm',annot=True)
```

Distribution of age, faceted by sex

```python
g = sns.FacetGrid(data=titanic,col='sex')
g.map(sns.histplot, 'age') 
```
