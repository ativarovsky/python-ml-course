# Pandas 3: Hierarchies & Indexing

```python
import numpy as np
import pandas as pd
from numpy.random import randn
```

### Create hierarchy

```python
# don't need to understand this for now
outside = ['G1', 'G1', 'G1', 'G2', 'G2', 'G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
```

```python
df = pd.DataFrame(randn(6,2), hier_index, ['A','B'])
```

```python
df
```

```python
df.loc['G1']
```

```python
# go further in the hierarchy
df.loc['G1'].loc[1] # always start from the outside and go in
```

Now label indices

```python
df.index.names = ['Groups', 'Num']
```

```python
df
```

```python
# say you want G2, 2, B
df.loc['G2'].loc[2]['B']
```

```python
# try another
df.loc['G1'].loc[3]['B']
```
