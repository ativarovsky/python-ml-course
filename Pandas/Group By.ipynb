# Group By

```python
import numpy as np
import pandas as pd
```

```python
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB', 'FB'],
        'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
        'Sales':[200, 120, 340, 124, 243, 350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.groupby('Company')
```

```python
# store as variable
byComp = df.groupby('Company')
```

```python
byComp
```

```python
# call an aggregate function
byComp.mean('Company')
```

```python
byComp.std()
```

```python
# say you want the sales sum of Facebook
byComp.mean().loc['FB']
```

```python
# note that you don't need to do this separately
df.groupby('Company').mean()
```

can also do max, min, etc.
