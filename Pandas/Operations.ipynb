## Select Data (using conditional selection)

```python
df['col1']>1 # just a boolean
```

```python
df[df['col1']>1]
```

### Apply Method 

```python
def times2(x) : 
    return(x*2)
```

```python
df['col1'].apply(times2)
```

Or for shorthand, use lambda function

```python
df['col1'].apply(lambda x: x*2)
```

```python
# how to choose columns and index
df.columns # note that this is an attribute, not a function, so doesn't need parentheses
```

```python
df.index # note that this is also an attribute
```

### Pivot Tables

```python
data = {'A':['foo','foo','foo','bar','bar','bar'],
     'B':['one','one','two','two','one','one'],
       'C':['x','y','x','y','x','y'],
       'D':[1,3,2,5,4,1]}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.pivot_table(values = 'D', index = ['A', 'B'], columns = ['C'])
```
