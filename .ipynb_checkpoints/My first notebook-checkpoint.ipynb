## List Comprehension

List comprehension is a different way to make for loops

```python
x = [1,2,3,4]
x
```

```python
out = []

for num in x:
    out.append(num**2)
```

```python
print(out)
```

```python
out = [num**2 for num in x]
```

## Lambda Expression 

```python
lambda var:var*2 # shortcut to using functions, the object before the colon 
# is what is fed in and the object on the other side is what is output
```

```python
seq = [1,2,3,4,5]
```

```python
output = map(lambda var:var*2, seq)
output
```

```python
output = list(map(lambda var:var*2, seq))
output
```

** Create a basic function that returns True if the word 'dog' is contained in the input string. Don't worry about edge cases like a punctuation being attached to the word dog, but do account for capitalization. **

```python
string = 'I found a DOG'
string

string_split = string.split()
string_split
```

Function: 
1. split string using .split()
2. check if each element of the .split() list is equal to "dog" using "in"

```python
"dog" in string_split
```

```python
def dogFunc(x):
    return "dog" in x.lower().split()
```

```python
"dog" in string.split()
```

```python
dogFunc(string)
```

** Create a function that counts the number of times the word "dog" occurs in a string. Again ignore edge cases. **

```python
string = "i love my dog and everyone else's dog"
string.split()
```

```python
string.split()[1]
```

```python
def dogCount(string):
    count = 0
    for i in string.split():
        if i.lower() == "dog": count = count + 1
#    if string.split()[i] == "dog": count = count + 1
    return count
```

```python
dogCount(string)
```

### Final Problem
**You are driving a little too fast, and a police officer stops you. Write a function
  to return one of 3 possible results: "No ticket", "Small ticket", or "Big Ticket". 
  If your speed is 60 or less, the result is "No Ticket". If speed is between 61 
  and 80 inclusive, the result is "Small Ticket". If speed is 81 or more, the result is "Big    Ticket". Unless it is your birthday (encoded as a boolean value in the parameters of the function) -- on your birthday, your speed can be 5 higher in all 
  cases. **

```python
def caught_speeding(speed, is_birthday:bool):
    # deal with birthday bool 
    if is_birthday: 
        birthday_allow = 5
    else: 
        birthday_allow = 0
    # speed with built in birthday allowance
    if speed <= 60 + birthday_allow:
        print("No Ticket")
    elif speed >= 61+birthday_allow and speed <= 80+birthday_allow:
        print("Small Ticket")
    else: 
        print("Big Ticket")
```

```python
caught_speeding(61, True)
```
