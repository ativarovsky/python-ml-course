## NumPy Arrays

Casting a list as a NumPy array

```python
my_list = [1,2,3]
```

```python
my_list
```

```python
import numpy as np
```

```python
arr = np.array(my_list)
```

```python
arr
```

```python
my_mat = [[1,2,3], [4,5,6], [7,8,9]]
```

```python
my_mat
```

```python
np.array(my_mat)
```

Programmatic way to create an array: 

```python
np.arange(0,11,2) # same as Python range function, has start, stop and interval arguments 
```

```python
np.zeros(3)
```

```python
np.zeros((5,5))
```

```python
np.zeros((2,3))
```

```python
np.ones(4) # one dimensional array
```

```python
np.ones((3,4)) # two dimensional array
```

```python
np.linspace(0,5,10)
```

Identity matrix 

```python
np.eye(4)
```

Arrays of random numbers

```python
np.random.rand(5)
```

```python
np.random.rand(5,5) #5-dimensional matrix of random numbers
```

Random numbers from a standard normal distribution

```python
np.random.randn(4,4)
```

```python
np.random.randint(1, 100) # exclusive of 100, inclusive of 1
```

```python
np.random.randint(1,100,10)
```

```python
arr= np.arange(25)
```

```python
arr
```

```python
ranarr = np.random.randint(0,50,10)
```

```python
ranarr
```

```python
arr.reshape(5,5) #only works if dimensions match 
```

What are the minimum and maximum values in `ranarr`?

```python
ranarr.min()
ranarr.max()
```

What are the indices of the minimum and maximum values in `ranarr`? 

```python
ranarr.argmax()
ranarr.argmin()
```

```python
arr = arr.reshape(5,5)
```
