# Matplot Lib

```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
```

```python
x = np.linspace(0,5,11)
y = x**2
```

```python
x
```

```python
y
```

## Functional Way 
(probably won't use)

```python
# FUNCTIONAL 
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y label')
plt.title('Title')
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r')
```

```python
plt.subplot(1,2,2)
plt.plot(y,x,'b')
```

## Object Oriented Way 
(more useful)

```python
data = np.arange(10)
```

```python
data
```

```python
plt.plot(data)
```

```python
# Object oriented 
fig = plt.figure()

axes = fig.add_axes([0.1,0.1,0.8,0.8]) # first two numbers are where the plot is in relation to
# the blank canvas, second two numbers are how much of the canvas it takes up

axes.plot(x,y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title('Set Title')
```

```python
fig = plt.figure() # create figure object from empty canvas

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # add axes to it
axes2 = fig.add_axes([0.2, 0.5, 0.4, 0.3])

axes1.plot(x,y) # add the plot
axes1.set_title('LARGER PLOT') # add title to it

axes2.plot(y,x)
axes2.set_title('SMALLER PLOT')
```

## Subplots

```python
fig,axes = plt.subplots(nrows = 1, ncols = 2)

axes[0].plot(x,y)
axes[0].set_title('First Plot')

axes[1].plot(x,y)
axes[1].set_title('Second Plot')
```

## Figure Size and DPI

```python
fig = plt.figure(figsize=(8,2))

ax = fig.add_axes([0,0,1,1])
ax.plot(x,x**2, label = 'X Squared')
ax.plot(x, x**3, label = 'X Cubed')

# add a legend
ax.legend(loc=0)
```

```python
fix,axes = plt.subplots(nrows=2,ncols=1,figsize=(8,2))
axes[0].plot(x,y)
axes[1].plot(x,y)

plt.tight_layout()
```
