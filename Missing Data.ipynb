# Pandas - Missing Data

```python
import numpy as np
import pandas as pd
```

```python
# create dictionary
d = {'A': [1,2,np.nan], 'B':[5,np.nan,np.nan], 'C':[1,2,3]}
```

```python
d
```

```python
df = pd.DataFrame(d)
```

```python
df
```

```python
df.dropna() # pandas drops any row with one or more NA's
```

```python
df.dropna(axis = 1) # if you set axis = 1 it'll drop columns of null values
```

```python
# you can also set a threshold for how many NAs before it drops
df.dropna(thresh = 2)
```

### Filling In Missing Values

```python
df.fillna(value = 'FILL VALUE')
```

```python
df['A'].fillna(value = df['A'].mean())
```
