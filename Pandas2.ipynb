## Conditional Selection 

```python
df > 0 # same as NumPy
```

to filter on this bool, use df[df...]] notation 

```python
booldf = df>0
```

```python
booldf
```

```python
df[booldf]
```

Just a column: 

```python
df['W'] > 0 
```

Filter based on where W is greater than 0

```python
df[df['W']>0]
```

Another example 

```python
df[df['Z']<0]
```

go back to W >0 

```python
resultdf = df[df['W']>0]
```

```python
resultdf
```

```python
resultdf['X']
```

What if you want just a few columns where the condition is true?

```python
df
```

```python
df[df['W']>0][['Y','X']]
```

```python
# the previous operation is the same as: 
boolser = df['W']>0
result = df[boolser]
mycols = ['Y', 'X']
result[mycols]
```

### Multiple Conditions

Pandas uses & (not 'and') and | (not 'or')

```python
df[(df['W']>0) & (df['Y']>0)] 
```

## Setting and Resetting Indexes

```python
# if you want to set this new variable as an index
df.set_index('States') # again, to fix this into the df, use inplace = True
```
