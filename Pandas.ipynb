# Pandas

## Series

```python
import numpy as np
```

```python
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_data = [10, 20, 30]
arr = np.array(my_data)
d = {'a': 10, 'b': 20, 'c': 30} 
```

Create a series, which is just a vector with an index (like a tracking system)

```python
pd.Series(data = my_data, index = labels) 
```

Notice the difference between using labels as index vs not. 

```python
pd.Series(arr)
```

```python
pd.Series(arr, index = labels)
```

Pass in a dictionary

```python
pd.Series(d)
```

You can put any data type in a series (even functions): 

```python
pd.Series(data = [sum, print, len])
```

```python
ser1 = pd.Series([1,2,3,4], ['USA', 'Germany', 'USSR', 'Japan']) # here the country names are indices and the numbers are the data
```

```python
ser1
```

```python
ser2 = pd.Series([1,2,3,4], ['USA', 'Germany', 'Italy', 'Japan'])
```

```python
ser2
```

```python
ser1['USA']
```

```python
ser3 = pd.Series(data = labels)
```

```python
ser3
```

```python
ser1 + ser2 #note that it adds what it can, where it can't find a match, it NA
```

## Dataframes

```python
from numpy.random import randn
```

```python
np.random.seed(101) 
```

Produce a 5 row, 4 column dataframe

```python
df = pd.DataFrame(randn(5,4))
```

```python
df
```

Now assign it column and row labels

```python
df = pd.DataFrame(randn(5,4), ['A', 'B', 'C', 'D', 'E'], ['W', 'X', 'Y', 'Z'])
```

```python
df
```

note that this is just a group of series; 
What if we want to grab a column from the data frame? Always use bracket notation 

```python
df['W']
```

```python
df[['W', 'Z']]
```

### Creating new columns/deleting columns

```python
df['new'] = df['W'] + df['Y']
```

```python
df
```

Know that rows are 0 as far as their axis and columns are 1, and default is axis = 0, so if you're trying to drop a column, you have to specify axis = 1

```python
df.drop('new', axis = 1)
```
